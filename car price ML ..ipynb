```python
# Pie chart
```

```python
df_1.columns
```

```python
df_2 = df_1[["Mileage","Price"]]
```

```python
df_2.describe()
```

```python
overall_investment = {i:df_2[i].sum() for i in df_2.columns}
```

```python
overall_investment
```

```python
overall_investment.values()
```

```python
overall_investment.keys()
```

```python
plt.figure(figsize=(20,7))

plt.pie(overall_investment.values(),  labels=overall_investment.keys() , autopct= '%1.3f%%')

plt.title("Pie Plot")
plt.show()
```

```python
#Donut Chart
```

```python
plt.figure(figsize=(20,7))

plt.pie(overall_investment.values(),  labels=overall_investment.keys() , autopct= '%1.3f%%')

circle = plt.Circle((0,0), 0.90, fc = 'white')

donut = plt.gcf()
donut.gca().add_artist(circle)


plt.title("Pie Plot")
plt.show()
```
